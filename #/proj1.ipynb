```python
# Validate and convert date columns

date_format = "%m/%d/%Y %H:%M:%S"

datasets = {
    "Logon": logon,
    "Device": device,
    "File": file
}

for name, df in datasets.items():

    # Keep the original values for validation
    original_dates = df["date"].copy()

    # Convert dates using the explicit format
    converted_dates = pd.to_datetime(
        original_dates,
        format=date_format,
        errors="coerce"
    )

    # Identify invalid dates
    invalid_dates = converted_dates.isna()

    print(f"\n--- {name} Dataset ---")
    print("Total records:", len(df))
    print("Valid dates:", converted_dates.notna().sum())
    print("Invalid dates:", invalid_dates.sum())

    # Display invalid values if they exist
    if invalid_dates.any():
        print("Examples of invalid dates:")
        print(original_dates[invalid_dates].head(10).to_list())

    else:
        # Update the date column only when all dates are valid
        df["date"] = converted_dates

        print("Date conversion successful!")
        print("Minimum date:", df["date"].min())
        print("Maximum date:", df["date"].max())
```

```python
# Convert data columns to datatime format
logon['date'] = pd.to_datetime(logon['date'])
device['date'] = pd.to_datetime(device['date'])
file['date'] = pd.to_datetime(file['date'])
logon.info()
```

```python
#Extract the hour from each activity timestamp
logon['hour'] = logon['date'].dt.hour
device['hour'] = device['date'].dt.hour
file['hour'] = file['date'].dt.hour
logon[['date', 'hour']].head()
```

```python
# Identify activities outside normal working hours
logon['after_hours'] = ((logon['hour'] < 8) | (logon['hour'] > 17)).astype(int)

device['after_hours'] = ((device['hour'] < 8) | (device['hour'] > 17)).astype(int)

file['after_hours'] = ((file['hour'] < 8) | (file['hour'] > 17)).astype(int)
logon[['hour', 'after_hours']].head()
```

Feature Engineering

```python
# Generate behavioral features for ecah user
logon_features = logon.groupby('user').agg(
    logon_count=('activity', 'count'),
    unique_pcs=('pc', 'nunique'),
    after_hours_logons=('after_hours', 'sum')
)
after_hours_logons = logon_features['after_hours_logons'].mean()
logon_features['after_hours_logons_rate'] = logon_features['after_hours_logons'] / after_hours_logons


logon_features.head()
```

```python
# Duration
```

```python
# Generate device activity features for ecah user
device_features = device.groupby('user').agg(
    device_events=('activity', 'count'),
    device_unique_pcs=('pc', 'nunique'),
    after_hours_device=('after_hours', 'sum')
)

device_features.head()
```

```python
# Generate file activity features for ecah user
file_features = file.groupby('user').agg(
    file_events=('filename', 'count'),
    unique_file=('filename', 'nunique'),
    after_hours_file=('after_hours', 'sum')
)

file_features.head()
```

```python
# Marge all user behavioral features
user_features = logon_features.join(device_features, how='outer')
user_features = user_features.join(file_features, how='outer')

user_features = user_features.fillna(0)
user_features.head()
```

```python
# Enhanced Behavioral Feature Engineering (including weekends)
logon['is_weekend'] = logon['date'].dt.dayofweek.isin([5, 6]).astype(int)

logon_features = logon.groupby('user').agg(
    logon_count=('activity', 'count'),
    unique_pcs=('pc', 'nunique'),
    after_hours_logons=('after_hours', 'sum'),
    weekend_logons=('is_weekend', 'sum')
)

# Enhanced Device Features (separating Connect and Disconnect)
device['is_weekend'] = device['date'].dt.dayofweek.isin([5, 6]).astype(int)
device_connects = device[device['activity'] == 'Connect'].groupby('user').size().rename('device_connects')
device_disconnects = device[device['activity'] == 'Disconnect'].groupby('user').size().rename('device_disconnects')

device_base = device.groupby('user').agg(
    device_events=('activity', 'count'),
    device_unique_pcs=('pc', 'nunique'),
    after_hours_device=('after_hours', 'sum'),
    weekend_device=('is_weekend', 'sum')
)
device_features = pd.concat([device_base, device_connects, device_disconnects], axis=1).fillna(0)

# Enhanced File Features
file['is_weekend'] = file['date'].dt.dayofweek.isin([5, 6]).astype(int)

file_features = file.groupby('user').agg(
    file_events=('filename', 'count'),
    unique_file=('filename', 'nunique'),
    after_hours_file=('after_hours', 'sum'),
    weekend_file=('is_weekend', 'sum')
)

# Merge all enhanced features
user_features = logon_features.join(device_features, how='outer')
user_features = user_features.join(file_features, how='outer')
user_features = user_features.fillna(0)

print("Enhanced feature matrix shape:", user_features.shape)
print("Features created:", list(user_features.columns))
```

# Model Evaluation for feature columns including the new behavioral features

```python
# Select only the original behavioral features
feature_columns = [
    'logon_count',
    'unique_pcs',
    'after_hours_logons',
    'weekend_logons',
    'device_events',
    'device_unique_pcs',
    'after_hours_device',
    'weekend_device',
    'device_connects',
    'device_disconnects',
    'file_events',
    'unique_file',
    'after_hours_file',
    'weekend_file'
]
X = user_features[feature_columns].copy()

# Check for missing and infinite values
print("Missing values:", X.isnull().sum().sum())
print("Infinite values:", np.isinf(X).sum().sum())

# Train the Isolation Forest model
model = IsolationForest(
    contamination=0.03,
    random_state=42
)

model.fit(X)

# Generate predictions
predictions = model.predict(X)

# Convert predictions into binary labels
user_features['anomaly'] = (
    predictions == -1
).astype(int)

# Calculate anomaly scores
# Higher scores indicate more unusual behavior
user_features['anomaly_score'] = (
    -model.decision_function(X)
)

# Display prediction counts
print("\nAnomaly Distribution:")
print(user_features['anomaly'].value_counts())

# Calculate anomaly percentages
anomaly_percentage = (
    user_features['anomaly'].mean() * 100
)

print("\nAnomaly Percentage:",
      round(anomaly_percentage, 2), "%")

# Display the top 10 most anomalous users
top_anomalies = user_features.sort_values(
    by='anomaly_score',
    ascending=False
)

print("\nTop 10 Most Anomalous Users:")

print(
    top_anomalies[
        feature_columns + ['anomaly', 'anomaly_score']
    ].head(10)
)

# Display anomaly score statistics
print("\nAnomaly Score Statistics:")
print(user_features['anomaly_score'].describe())
```

```python
# Compare a baseline feature set with the full behavioral feature set
baseline_columns = [
    'logon_count', 'unique_pcs', 'after_hours_logons',
    'device_events', 'device_unique_pcs', 'after_hours_device',
    'file_events', 'unique_file', 'after_hours_file'
]

enhanced_columns = [
    'logon_count', 'unique_pcs', 'after_hours_logons', 'weekend_logons',
    'device_events', 'device_unique_pcs', 'after_hours_device',
    'weekend_device', 'device_connects', 'device_disconnects',
    'file_events', 'unique_file', 'after_hours_file', 'weekend_file'
]

train_idx, test_idx = train_test_split(
    np.arange(len(user_features)),
    test_size=0.2,
    random_state=42
)

comparison_rows = {}

for name, columns in [
    ('Baseline', baseline_columns),
    ('Enhanced', enhanced_columns)
]:
    X_compare = user_features[columns].fillna(0)

    comparison_model = IsolationForest(
        contamination=0.03,
        random_state=42
    )

    start = time.perf_counter()
    comparison_model.fit(X_compare.iloc[train_idx])
    fit_seconds = time.perf_counter() - start

    train_predictions = comparison_model.predict(X_compare.iloc[train_idx])
    test_predictions = comparison_model.predict(X_compare.iloc[test_idx])
    test_scores = -comparison_model.decision_function(X_compare.iloc[test_idx])

    comparison_rows[name] = {
        'fit_seconds': fit_seconds,
        'train_anomaly_rate_%': (train_predictions == -1).mean() * 100,
        'test_anomaly_rate_%': (test_predictions == -1).mean() * 100,
        'mean_test_anomaly_score': test_scores.mean(),
        'max_test_anomaly_score': test_scores.max()
    }

comparison = pd.DataFrame.from_dict(comparison_rows, orient='index')
metrics = comparison.copy()
positions = np.arange(len(metrics))
bar_width = 0.2
colors = ['steelblue', 'green']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Fit time
axes[0, 0].bar(positions, metrics['fit_seconds'], width=bar_width, color=colors)
axes[0, 0].set_title('Model Fit Time')
axes[0, 0].set_ylabel('Seconds')
axes[0, 0].set_xticks(positions, metrics.index, rotation=0)
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.4)

# Train vs. test anomaly rates
group_width = 0.25
axes[0, 1].bar(
    positions - group_width / 2,
    metrics['train_anomaly_rate_%'],
    width=group_width,
    color='steelblue',
    label='Train'
)
axes[0, 1].bar(
    positions + group_width / 2,
    metrics['test_anomaly_rate_%'],
    width=group_width,
    color='green',
    label='Test'
)
axes[0, 1].set_title('Anomaly Rate: Train vs. Test')
axes[0, 1].set_ylabel('Anomaly rate (%)')
axes[0, 1].set_xticks(positions, metrics.index, rotation=0)
axes[0, 1].legend()
axes[0, 1].grid(axis='y', linestyle='--', alpha=0.4)

# Mean test anomaly score
group_width = 0.20
axes[1, 0].bar( 
    positions, metrics['mean_test_anomaly_score'],
    width=bar_width, color=colors
)
axes[1, 0].set_title('Mean Test Anomaly Score')
axes[1, 0].set_ylabel('Score')
axes[1, 0].set_xticks(positions, metrics.index, rotation=0)
axes[1, 0].axhline(0, color='black', linewidth=0.6)
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.4)

# Maximum test anomaly score
axes[1, 1].bar(
    positions, metrics['max_test_anomaly_score'],
    width=bar_width, color=colors
)
axes[1, 1].set_title('Maximum Test Anomaly Score')
axes[1, 1].set_ylabel('Score')
axes[1, 1].set_xticks(positions, metrics.index, rotation=0)
axes[1, 1].grid(axis='y', linestyle='--', alpha=0.4)

fig.suptitle('Isolation Forest Model Comparison', fontsize=14)
fig.tight_layout()
fig.savefig('model_comparison.png', dpi=300, bbox_inches='tight')
plt.show()


print("Chart saved to model_comparison.png")
```

## Unsupervised Detector Results

The following comparison fits Isolation Forest and One-Class SVM on the training users and reports flags on the held-out users. Because this project has no verified threat labels, these are anomaly rates and detector agreement—not accuracy, Precision, Recall, or F1. Random Forest and XGBoost are supervised classifiers and are not evaluated here without labels.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.svm import OneClassSVM

X_train_unsup = X.iloc[train_idx].copy()
X_test_unsup = X.iloc[test_idx].copy()

if len(X_train_unsup) < 2 or len(X_test_unsup) == 0:
    raise ValueError("The train/test split is too small for unsupervised model comparison.")
if not np.isfinite(X_train_unsup.to_numpy()).all() or not np.isfinite(X_test_unsup.to_numpy()).all():
    raise ValueError("The feature data contains missing or infinite values.")

isolation_forest_eval = IsolationForest(
    contamination=0.03,
    random_state=42
)
start = time.perf_counter()
isolation_forest_eval.fit(X_train_unsup)
if_fit_seconds = time.perf_counter() - start
if_test_flags = (isolation_forest_eval.predict(X_test_unsup) == -1)

scaler_eval = StandardScaler()
X_train_scaled = scaler_eval.fit_transform(X_train_unsup)
X_test_scaled = scaler_eval.transform(X_test_unsup)
one_class_svm_eval = OneClassSVM(
    nu=0.03,
    kernel='rbf',
    gamma='scale'
)
start = time.perf_counter()
one_class_svm_eval.fit(X_train_scaled)
svm_fit_seconds = time.perf_counter() - start
svm_test_flags = (one_class_svm_eval.predict(X_test_scaled) == -1)

unsupervised_results = pd.DataFrame([
    {
        'Model': 'Isolation Forest',
        'Test profiles flagged': int(if_test_flags.sum()),
        'Test anomaly rate (%)': float(if_test_flags.mean() * 100),
        'Fit time (seconds)': if_fit_seconds
    },
    {
        'Model': 'One-Class SVM',
        'Test profiles flagged': int(svm_test_flags.sum()),
        'Test anomaly rate (%)': float(svm_test_flags.mean() * 100),
        'Fit time (seconds)': svm_fit_seconds
    }
])

display(unsupervised_results.round(3))

flag_overlap = int((if_test_flags & svm_test_flags).sum())
flag_union = int((if_test_flags | svm_test_flags).sum())
agreement_jaccard = flag_overlap / flag_union if flag_union else 1.0
print("Users flagged by both models:", flag_overlap)
print("Jaccard overlap of flagged users:", round(agreement_jaccard, 3))

unsupervised_results.to_csv(
    'Results/unsupervised_model_comparison.csv',
    index=False
)
unsupervised_predictions = pd.DataFrame({
    'isolation_forest_anomaly': if_test_flags.astype(int),
    'one_class_svm_anomaly': svm_test_flags.astype(int)
}, index=X_test_unsup.index)
unsupervised_predictions.to_csv(
    'Results/unsupervised_model_predictions.csv',
    index_label='user'
)
print("Saved unlabeled comparison and test predictions under Results/.")
```

## XGBoost: Isolation Forest Pseudo-label Demonstration

There are no verified threat labels in the supplied data. This trains XGBoost to imitate Isolation Forest labels generated from the training split, then measures agreement with Isolation Forest on the held-out split. These scores are **not** threat-detection Precision, Recall, or F1 and must not be presented as such.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from xgboost import XGBClassifier

teacher = IsolationForest(
    contamination=0.03,
    random_state=42
)
teacher.fit(X.iloc[train_idx])
y_train_pseudo = (teacher.predict(X.iloc[train_idx]) == -1).astype(int)
y_test_pseudo = (teacher.predict(X.iloc[test_idx]) == -1).astype(int)

positive_count = int(y_train_pseudo.sum())
if positive_count == 0 or positive_count == len(y_train_pseudo):
    raise ValueError("Isolation Forest produced only one class for XGBoost training.")

xgboost_model = XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=(len(y_train_pseudo) - positive_count) / positive_count,
    random_state=42,
    n_jobs=1,
    tree_method='hist'
)

start = time.perf_counter()
xgboost_model.fit(X.iloc[train_idx], y_train_pseudo)
xgboost_fit_seconds = time.perf_counter() - start

xgboost_test_predictions = xgboost_model.predict(X.iloc[test_idx])
xgboost_test_probabilities = xgboost_model.predict_proba(X.iloc[test_idx])[:, 1]
pseudo_label_auc = (
    roc_auc_score(y_test_pseudo, xgboost_test_probabilities)
    if len(np.unique(y_test_pseudo)) == 2
    else np.nan
)

xgboost_pseudo_evaluation = pd.DataFrame([{
    'Model': 'XGBoost',
    'Reference': 'Isolation Forest pseudo-labels',
    'Accuracy (agreement)': accuracy_score(y_test_pseudo, xgboost_test_predictions),
    'Precision (agreement)': precision_score(y_test_pseudo, xgboost_test_predictions, zero_division=0),
    'Recall (agreement)': recall_score(y_test_pseudo, xgboost_test_predictions, zero_division=0),
    'F1 (agreement)': f1_score(y_test_pseudo, xgboost_test_predictions, zero_division=0),
    'ROC AUC (agreement)': pseudo_label_auc,
    'Fit time (seconds)': xgboost_fit_seconds
}])

display(xgboost_pseudo_evaluation.round(3))
print("These metrics measure agreement with Isolation Forest, not verified threat detection.")

xgboost_pseudo_evaluation.to_csv(
    'Results/xgboost_pseudo_label_evaluation.csv',
    index=False
)
xgboost_pseudo_predictions = pd.DataFrame({
    'isolation_forest_pseudo_label': y_test_pseudo,
    'xgboost_prediction': xgboost_test_predictions.astype(int),
    'xgboost_anomaly_probability': xgboost_test_probabilities
}, index=X.iloc[test_idx].index)
xgboost_pseudo_predictions.to_csv(
    'Results/xgboost_pseudo_label_predictions.csv',
    index_label='user'
)
print("Saved XGBoost pseudo-label agreement results under Results/.")
```
